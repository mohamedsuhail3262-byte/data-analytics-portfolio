# Retail Sales Analysis

## Project Overview

This project analyzes retail sales data to understand sales performance, profitability, customer behavior, product performance, regional trends, and order activity.

The analysis uses Python libraries such as Pandas, NumPy, Matplotlib, and Seaborn to perform data analysis and create meaningful visualizations for business decision-making.

## Objectives

- Analyze overall sales and profit performance.
- Identify high-performing products and departments.
- Understand customer purchasing behavior.
- Compare regional sales and profitability.
- Analyze order status and transaction trends.
- Identify important relationships between numerical variables.
- Generate actionable business insights and recommendations.

```python
df[['Sales','Cost','Profit','Quantity']].describe()
```

```python
category_analysis=df.groupby('Department')[['Sales','Cost','Profit','Quantity']].sum().sort_values('Sales',ascending=False)
```

## 1. Sales by Department

```python
plt.figure(figsize=(10, 6))

sales = category_analysis['Sales'].sort_values(ascending=False)

sales.plot(kind='bar',edgecolor='black')

plt.title('Sales by Department', fontsize=14, fontweight='bold')
plt.xlabel('Department')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
top_10_products=(df.groupby('Product')['Sales'].sum().sort_values(ascending=False).head(10))

```

## 2. Top 10 Products by Sales

```python
plt.figure(figsize=(10, 6))

sales = top_10_products.sort_values()

plt.hlines(y=sales.index,xmin=0,xmax=sales.values,linewidth=2)

plt.plot(sales.values,sales.index,'o',markersize=8)

plt.title('Top 10 Products by Sales',fontsize=14,fontweight='bold')

plt.xlabel('Total Sales')
plt.ylabel('Product')

plt.grid(axis='x',linestyle='--',alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
monthly_sales=df.groupby(df['Date'].dt.to_period('M'))['Sales'].sum()
```

## 3. Monthly Sales Trend

```python
plt.figure(figsize=(12, 5))

monthly_sales.plot(kind='line',marker='o',linewidth=2)

plt.title('Monthly Sales Trend', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
monthly_profit=df.groupby(df['Date'].dt.to_period('M'))['Profit'].sum()
```

## 4. Monthly Profit Trend

```python
plt.figure(figsize=(12, 5))

monthly_profit.plot(kind='line',marker='o',linewidth=2)

plt.title('Monthly Profit Trend', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Total Profit')

plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```

## Key Business Insights

- Electronics is the strongest-performing department in terms of overall sales and profit.
- Smartphone and Tablet are among the highest-profit products.
- Sales and Profit show a strong positive relationship, indicating that higher sales generally correspond to higher profit.
- Returning customers contribute a larger share of total profit than new customers.
- Sales and profit performance varies across regions, highlighting differences in regional performance.
- Monthly transaction volume fluctuates throughout the year, indicating changes in customer purchasing activity.
- Completed orders account for the majority of orders, while cancelled and returned orders represent smaller shares.
- Electronics has the highest average profit per transaction among the departments.
- Profit is concentrated around the lower range, with some high-profit transactions creating a right-skewed distribution.
- The correlation analysis shows strong relationships among key financial variables such as Sales, Cost, and Profit.

### Conclusion

The analysis provides a clear understanding of sales performance, profitability, customer behavior, product performance, regional performance, and order activity. The findings highlight the key areas contributing to overall business performance and support data-driven decision-making.
